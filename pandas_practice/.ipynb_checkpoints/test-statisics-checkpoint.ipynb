## Z-Score
A Z-Score tells me how far my result is from the mean while a p-value tells me how rare the result is if the null hypothesis is true. A big z-score leads to a small p-value while a small z-score leads to a large p-value. They have a negative correlation. 

* If p < alpha, z will be beyond the critical value
* If p > alpha, z will be inside

Note if you z-score is positive, you do 1-area to the left to get the score and multiply by two. 

## p_value < alpha
p-value (the probability of the result under the null) says that my result is so rare that it is safe to reject the null hypothesis.

## Differences between left and right tail

* Left tail test: used if alternative hypothesis is less than null.
* Right-tail test: used if alternative hypothesis is greater than null
* Two-tail test: used if alternative hypothesis is not equal to the null

# Newsletter Signup Rates (A/B Test)
You work for a website that’s testing two different landing pages (A and B) to see which one drives more newsletter signups.
Page A: Out of 300 visitors, 42 signed up for the newsletter.
Page B: Out of 320 visitors, 56 signed up for the newsletter.

## Question:
Is there a statistically significant difference in the signup rates between the two pages at a 0.05 significance level?

#### Hypotheses
* **Null:** pageA = pageB
* **Alternative:** = pageA - pageB > 0
* **Z-Critical** = 1.96
* **alpha** = .05

```python
phatA = 42/300
phatB = 56/320
```

```python
pooled_p = (42 + 56) / (300 + 320)
pooled_p
```

```python
SE = math.sqrt((pooled_p * (1-pooled_p)) * ((1/300) + (1/320)))
SE
```

```python
z = abs((phatA - phatB))/ SE
z
p_value = 2* (1 - .8830) #wanted to the right of the p-value
print(f"Z-Score: {z}\nP-Value: {p_value}")
```

## Fail to reject the null hypothesis
The P-Value is .234 and the z-score is 1.19. And there is no significant difference in the signup rates between the two.  

# Vaccine Efficacy in Two Cities
A new flu vaccine was rolled out in two cities. After flu season:
* City 1: Out of 800 vaccinated people, 40 still caught the flu.
* City 2: Out of 900 vaccinated people, 27 caught the flu.

## Question:
Is there evidence that the vaccine worked better in one city compared to the other? Use a 2-proportion z-test at a 0.05 significance level.

* **Null Hypothesis:** city1 - city2 = 0
* **Alternate Hypothesis:** city1 - city2 != 0
* **Z-Critical:** 1.96
* **Alpha:** .05

```python
city1 = 40 / 800
city2 =  27/ 900
print(f"For City 1: {city1} percent still caught the flu\nFor City 2: {city2} percent still caught the flu")
```

```python
check = city1 - city2
check
```

### Which city did better?
In the population samples for CITY 2 and CITY 1, in CITY 2 3 percent of those vaccinated caught the flu while for CITY 1 5 percent of those vaccinated caught the flu. There is a 2 percent difference between those that caught the flu after vaccination in both cities.



```python
pooled_p = (40 + 27) / (800 + 900)
pooled_p
```

```python
SE = math.sqrt((pooled_p *(1-pooled_p)) * ((1/800) + (1/900)))
SE
```

```python
z = abs(city1 - city2)/SE
print(z)
p_value = 2*(1 - .9830)
p_value
print(f"Z-Score:{z}\nP-value:{p_value}")
```

```python
#testing the p_value using survival function
test_pv = 2* norm.sf(abs(z))
print(test_pv)
```

### Reject the null hypothesis
There is evidence that the vaccine worked in City 2 over City 1. Only 3 percent of the sample population caught the flu after vaccination in City 2. For City 1, 5 percent caught the flu after vaccination. 

* **Z-statistic** is 2.12 standard deviations above what is expected if the null hypothesis were true.
* **Standard Errors** With an SE of .01, I can be confident about the difference.
* **P-value** is 0.03, which is less than alpha. We have 3 percent chance of seeing a z-statistic greater than the null hypothesis. 

# Tech Support Ticket Success
> Your tech support team closes tickets with a success rate of 85%.

## Question:
> What is the probability that, out of the next 12 tickets, at least 10 will be closed successfully? (Calculate the probability using the binomial distribution.)

> Where X is the number of tickets successfully closed out 12.

* p( X >= 10) = P(X=10) + P(X=11) + p(X=12)
* where X ~ Binomial(n=12,p=0.85)

> The probability for combinations is P(X = k) = comb(n,k) * p^k) *((1-p)^(n-k))



```python
n = 12
p = .85
k = 10
#combinations for each outcome
comb10 = round((math.factorial(12))/(math.factorial(10) * math.factorial(12-10)), 3)
comb11 = round(math.factorial(12)/(math.factorial(11) *math.factorial(12-11)),3)
comb12 = round(math.factorial(12)/(math.factorial(12) * math.factorial(12-12)), 3)
p_of_X = comb10 + comb11 +comb12
print(f"Combinations 10: {comb10}\nCombinations 11: {comb11}\nCombinations 12: {comb12}")
#Probabilities for each outcome
p10 = comb10 * (p**10) * ((1-p)**2)
p11 = comb11 * (p**11) * ((1-p)**1)
p12 = comb12 * (p**12) * ((1-p)**0)
total_prob = p10 + p11 + p12
print(f"Probability: {total_prob:.3f}")
```

```python
#Checking the probability
prob_at_least_10 = binom.pmf(10,n,p) + binom.pmf(11,n,p) + binom.pmf(12,n,p)
print(f"Verifying that probability of at least 10 is {prob_at_least_10}")
survival_function = binom.sf(k-1, n, p)
print(f"The Survival Function: {survival_function}")
```

## Answer:
The tech team has a 74 percent chance of successfully closing at least 10 out of 12 tickets.

# Dating App Profile Pics (A/B Test)

> A dating app tests whether having a pet in your profile photo increases your match rate.

* **Group A (with pet):** 120 out of 500 users got at least one match.
* **Group B (no pet):** 85 out of 450 users got at least one match.

### Hypotheses

* **Null Hypothesis:** GroupA = GroupB
* **Alternate Hypothesis:** GroupA != GroupB
* **Z-Critical:** 1.96
* **Alpha:** 0.05

> We fail to reject the null hypothesis. There is a change between Group A and Group B, with a 5 percent difference between the two groups. Group A has a 24 percent proportion in success while Group B has a 19 percent proportion. Although placing a pet in the profile photo does have a 5 percent increase in match rate, it is not statistically significant at the 0.05 level.

```python
groupA = 120/500
groupB = 85/450
change = round(groupA - groupB, 2)
print(f"There's a {change} percent difference between Group A and Group B\n with {round(groupA,2)}"
        f" having a proportion of successes compared to Group B's {round(groupB, 2)} proportion of successes")
```

```python
pooled_phat = (120 + 85) / (500 +450)
se = math.sqrt(pooled_phat * (1-pooled_phat) * ((1/500)+(1/450)))
print(f"Pooled Propotion: {pooled_phat}")
print(f"Standard Error: {se}")
```

```python
z = round(((groupA - groupB) / se),3)
print(f"Z-Score: {z}")
p_value = 2 * .0281
print(f"P-Value is the right tail for the Z-Score in the Z-Table. This is {p_value}.")
```

```python
if p_value < alpha:
    print("Reject the null hypothesis")
else:
    print("Fail to reject the null hypothesis")
```
